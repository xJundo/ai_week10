# 2. Problem definition

- **Input**: 13 chemical measurements of a wine (alcohol, malic acid, ...)
- **Target**: the grape cultivar (`class_0`, `class_1`, `class_2`)
- **Problem type**: multiclass classification (3 classes)

```python
import pandas as pd
from sklearn.datasets import load_wine

wine = load_wine(as_frame=True)
df = wine.frame
print(df.shape)
df.head()
```

# 3. Evaluation metric

```python
counts = df["target"].value_counts().sort_index()
pd.DataFrame({
    "class": wine.target_names,
    "samples": counts.values,
    "ratio (%)": (counts / len(df) * 100).round(1).values,
}).set_index("class")
```

The classes are 59 : 71 : 48 (33% / 40% / 27%): mildly imbalanced, but no class is rare, so **accuracy** a fair metric and easy to read. We also look at the accuracy of each class separately, to check that a model is not good on one class only.

# 4. Common-sense baseline

Always answer with the most frequent class. It gives a reference since a real model is only useful if it clearly beats it.

Always predicting `class_1` gives about 40% accuracy.

```python
y = df["target"]
majority = y.mode()[0]
pred = pd.Series(majority, index=y.index)
baseline_acc = (pred == y).mean()
print(f"Most frequent class: {wine.target_names[majority]}")
print(f"Baseline accuracy: {baseline_acc:.3f}")
```

**Accuracy per class**: for each type of wine, how many did the baseline get right? It only knows `class_1`, so it is 100% right on `class_1` and 0% right on the others.

```python
per_class = pd.DataFrame({
    "class": wine.target_names,
    "wines": y.value_counts().sort_index().values,
    "correct": (pred == y).groupby(y).sum().values,
})
per_class["accuracy (%)"] = (per_class["correct"] / per_class["wines"] * 100).round(1)
per_class.set_index("class")
```

# 6. Train / final test split

**80% training / 20% final test.** With less than 1,000 samples we can skip the check-up test set. We only have 178 wines, so a check-up set (about 27 wines) would be too small and we would lose training data. We will choose our models with cross-validation instead.

The split is stratified (same class ratio in each set) and shuffled, because the data is sorted by class.

```python
from sklearn.model_selection import train_test_split

train_df, final_test_df = train_test_split(
    df, test_size=0.2, stratify=df["target"], shuffle=True, random_state=42
)
pd.DataFrame({
    "train": train_df["target"].value_counts().sort_index(),
    "final test": final_test_df["target"].value_counts().sort_index(),
}).set_index(wine.target_names).assign(
    train_ratio=lambda t: (t["train"] / len(train_df) * 100).round(1),
    test_ratio=lambda t: (t["final test"] / len(final_test_df) * 100).round(1),
)
```

```python
print("train:", train_df.shape, "| final test:", final_test_df.shape)
```

We save the final test set in a file and we don't touch it until the end. From now on we only load the training data.

```python
import os

os.makedirs("../../data", exist_ok=True)
train_df.to_csv("../../data/train.csv", index=False)
final_test_df.to_csv("../../data/final_test.csv", index=False)

train = pd.read_csv("../../data/train.csv")
print(train.shape)
train.head()
```

# 7. K-fold cross-validation

**K = 5.** We have 142 wines for training, so each validation fold has about 28 wines. With K = 10 a fold would have only 14 wines, so the scores would be too noisy. With K = 3 the models would train on only 2/3 of the data. 5 is a good compromise.

The K-fold is stratified and shuffled. From Week 12 we will compare all the models with the average accuracy over the 5 folds.

```python
from sklearn.model_selection import StratifiedKFold

X_train = train.drop(columns="target").to_numpy()
y_train = train["target"].to_numpy()

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rows = []
for i, (tr_idx, val_idx) in enumerate(kfold.split(X_train, y_train)):
    counts = pd.Series(y_train[val_idx]).value_counts().sort_index()
    rows.append({"fold": i + 1, "train": len(tr_idx), "validation": len(val_idx),
                 "val class_0": counts[0], "val class_1": counts[1], "val class_2": counts[2]})
pd.DataFrame(rows).set_index("fold")
```

# 8. Preprocessing

**Missing values.** There are none, so nothing to fill. (If there were, we would use the median of the training data, because outliers don't change it.)

```python
print("Missing values in the training data:", train.isna().sum().sum())
```

**Categorical features.** None: all 13 features are numeric, so no one-hot encoding. The target is already an integer (0, 1, 2), so we can use it with `sparse_categorical_crossentropy` in Keras.

**Normalization.** Like in the course, we put each feature to mean 0 and standard deviation 1. The features have very different scales (`proline` is around 750, `hue` around 1) and a neural network trains badly like this. The mean and std are computed on the **training data only**, and the final test set will be scaled later with these same values.

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
X_train_norm = (X_train - mean) / std
```

**Before and after** (training data):

```python
cols = train.drop(columns="target").columns
print("BEFORE")
display(pd.DataFrame(X_train, columns=cols).head().round(2))
print("AFTER")
display(pd.DataFrame(X_train_norm, columns=cols).head().round(2))
```

```python
pd.DataFrame({
    "mean before": X_train.mean(axis=0),
    "std before": X_train.std(axis=0),
    "mean after": X_train_norm.mean(axis=0),
    "std after": X_train_norm.std(axis=0),
}, index=cols).round(2) + 0
```

# 10. Simple baseline model

Before deep learning, we try a very simple model: a **logistic regression**. It is our first real baseline, to compare with the common-sense baseline of section 4.

From now on the mean and std are computed inside each fold, on the 4 training folds only. This fixes the small point we noticed in section 9.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

logreg_accs = []
for tr_idx, val_idx in kfold.split(X_train, y_train):
    model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
    model.fit(X_train[tr_idx], y_train[tr_idx])
    logreg_accs.append(model.score(X_train[val_idx], y_train[val_idx]))

print("accuracy per fold:", np.round(logreg_accs, 3))
print(f"mean: {np.mean(logreg_accs):.3f} (std {np.std(logreg_accs):.3f})")
print(f"common-sense baseline (week 10): {baseline_acc:.3f}")
```

Logistic regression gets **97.9%** on average (std 2.8%), against **39.9%** for the common-sense baseline. This is about 58 points higher, so the features really help to find the cultivar, and even a very simple model works well. The neural network will have to be at least close to this score to be useful.

# 11. First deep learning model

Our first model with Keras:
- 2 hidden `Dense(16, relu)` layers. We only have 142 wines, so we keep the model small to avoid overfitting. `relu` is the default choice in the course.
- Last layer `Dense(3, softmax)`: we have 3 classes, so we need 3 outputs. `softmax` gives 3 probabilities that sum to 1 (`sigmoid` with 1 unit is only for 2 classes).
- Loss `sparse_categorical_crossentropy`: multiclass problem, and our target is an integer (0, 1, 2), not a one-hot vector.
- `rmsprop` optimizer and `accuracy` metric, like in the course. 50 epochs and batch size 16.

```python
import keras
from keras import layers

def build_model():
    keras.utils.set_random_seed(42)
    model = keras.Sequential([
        keras.Input(shape=(13,)),
        layers.Dense(16, activation="relu"),
        layers.Dense(16, activation="relu"),
        layers.Dense(3, activation="softmax"),
    ])
    model.compile(optimizer="rmsprop",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def cross_validate(build_fn, epochs=50):
    accs, histories = [], []
    for tr_idx, val_idx in kfold.split(X_train, y_train):
        mu = X_train[tr_idx].mean(axis=0)
        sigma = X_train[tr_idx].std(axis=0)
        model = build_fn()
        h = model.fit((X_train[tr_idx] - mu) / sigma, y_train[tr_idx],
                      validation_data=((X_train[val_idx] - mu) / sigma, y_train[val_idx]),
                      epochs=epochs, batch_size=16, verbose=0)
        accs.append(h.history["val_accuracy"][-1])
        histories.append(h.history)
    return accs, histories
```

# 12. Cross-validation score and learning curves

```python
accs_1, hist_1 = cross_validate(build_model)
print("accuracy per fold:", np.round(accs_1, 3))
print(f"mean: {np.mean(accs_1):.3f} (std {np.std(accs_1):.3f})")
```

Model 1 gets **96.5%** on average (std 3.8%). This is a bit lower than logistic regression (97.9%), and the std is higher. Fold 1 is the worst one (89.7%), so we look at it.

```python
def plot_curves(h, title):
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))
    axes[0].plot(h["loss"], label="train")
    axes[0].plot(h["val_loss"], label="validation")
    axes[0].set_title("Loss")
    axes[1].plot(h["accuracy"], label="train")
    axes[1].plot(h["val_accuracy"], label="validation")
    axes[1].set_title("Accuracy")
    for ax in axes:
        ax.set_xlabel("epoch")
        ax.legend()
    plt.suptitle(title)
    plt.show()

plot_curves(hist_1[0], "Model 1, fold 1")
```

```python
h = hist_1[0]
print("best validation loss:", round(min(h["val_loss"]), 3), "at epoch", int(np.argmin(h["val_loss"])) + 1)
print("last validation loss:", round(h["val_loss"][-1], 3))
print("last train loss:", round(h["loss"][-1], 4))
```

**State of the model.** The training loss goes almost to 0 and the training accuracy to 100%. The validation loss goes down until about epoch 30 (0.24), then goes up again (0.31 at epoch 50). The model starts to **overfit**: it learns the 113 training wines by heart instead of learning something general. The validation accuracy is stable, but the loss shows that the model is getting too sure about its mistakes.

# 13. Changing the model

The validation loss goes up while the training loss keeps going down, which is overfitting. To fix it, we add a `Dropout(0.3)` after each hidden layer. During training, 30% of the neurons are randomly turned off at each step, so the model cannot rely on a few neurons and learns more general rules. Dropout is in the course (chapter 4). Everything else stays the same, so we can see the effect of this one change.

```python
def build_model_dropout():
    keras.utils.set_random_seed(42)
    model = keras.Sequential([
        keras.Input(shape=(13,)),
        layers.Dense(16, activation="relu"),
        layers.Dropout(0.3),
        layers.Dense(16, activation="relu"),
        layers.Dropout(0.3),
        layers.Dense(3, activation="softmax"),
    ])
    model.compile(optimizer="rmsprop",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model

accs_2, hist_2 = cross_validate(build_model_dropout)
print("accuracy per fold:", np.round(accs_2, 3))
print(f"mean: {np.mean(accs_2):.3f} (std {np.std(accs_2):.3f})")
```

```python
plot_curves(hist_2[0], "Model 2 (dropout), fold 1")
```

```python
h = hist_2[0]
print("best validation loss:", round(min(h["val_loss"]), 3), "at epoch", int(np.argmin(h["val_loss"])) + 1)
print("last validation loss:", round(h["val_loss"][-1], 3))
print("last train loss:", round(h["loss"][-1], 4))
```

With dropout the average accuracy goes from **96.5% to 97.2%** and the std from **3.8% to 1.4%**. On fold 1 the validation loss now keeps going down until the end (0.19 at epoch 50) and it no longer goes up again, so the overfitting is reduced. The train loss is a bit higher than the validation loss, because the dropout is only active during training. The curves are still going down at epoch 50, so the model could probably train a bit longer.

The gain in accuracy is small (about 1 wine in 142), so we should not over-read it. But the std is clearly lower, which means the model is more stable from one fold to another.

# 14. Model summary

```python
build_model_dropout().summary()
```

**Parameters of the first layer.** The input has 13 features and the layer has 16 units. Each unit has one weight per feature and one bias:

13 × 16 + 16 = **224 parameters**

The same way, the second layer has 16 × 16 + 16 = 272 parameters and the output layer has 16 × 3 + 3 = 51. The total is 224 + 272 + 51 = **547**. The Dropout layers have no parameter.
